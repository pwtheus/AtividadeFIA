# Rede neural de duas camadas

O código gera 64 exemplos aleatórios. Cada exemplo tem 1.000 valores de entrada e 10 valores que a rede tenta prever.

Para fazer a previsão, a rede combina as entradas com os pesos `w1` e aplica a função sigmoide. O resultado fica em `h`, que representa a camada escondida de 100 neurônios. Depois, a rede combina `h` com os pesos `w2` para obter a previsão `y_pred`.

A rede compara `y_pred` com os valores esperados `y`. A soma dos quadrados das diferenças é o erro `loss`. Em cada linha da saída, `print(t, loss)` mostra o número da rodada e o erro.

As linhas com `grad_` calculam como cada peso influencia o erro. Esse cálculo começa na saída e volta até a camada escondida, por isso é chamado de retropropagação. A rede usa esses gradientes para ajustar os pesos com uma taxa de aprendizado de 0,0001. O processo se repete 2.000 vezes.

```python
import numpy as np
from numpy.random import randn

N, D_in, H, D_out = 64, 1000, 100, 10
x, y = randn(N, D_in), randn(N, D_out)
w1, w2 = randn(D_in, H), randn(H, D_out)

for t in range(2000):
    h = 1 / (1 + np.exp(-x.dot(w1)))
    y_pred = h.dot(w2)
    loss = np.square(y_pred - y).sum()
    print(t, loss)

    grad_y_pred = 2.0 * (y_pred - y)
    grad_w2 = h.T.dot(grad_y_pred)
    grad_h = grad_y_pred.dot(w2.T)
    grad_w1 = x.T.dot(grad_h * h * (1 - h))

    w1 -= 1e-4 * grad_w1
    w2 -= 1e-4 * grad_w2
```
